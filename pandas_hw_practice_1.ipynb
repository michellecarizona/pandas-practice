```python
import pandas as pd
import numpy as np
```

loads Pandas and NumPy

```python
# Create a Series from a Python list
my_series = pd.Series([10, 20, 30, 40, 50])
print(my_series)
```

Converts a Python list into a Series. By default, Pandas assigns numbers as the index starting at 0.

```python
# Create a Series with custom labels
my_series = pd.Series(
    [10, 20, 30, 40, 50],
    index=["num1", "num2", "num3", "num4", "num5"]
)

print(my_series)
```

Assigns custom text labels to each element instead of standard numbers as the index.

```python
# Create a Series from a dictionary
my_dict = {
    "num1": 6,
    "num2": 7,
    "num3": 8
}

my_series = pd.Series(my_dict)
print(my_series)
```

Converts a Python dictionary into a Series. Dictionary keys automatically become the row labels (index), and values become the data.

```python
# Access values by label and position

my_series = pd.Series(
    [10, 20, 30, 40, 50],
    index=["num1", "num2", "num3", "num4", "num5"]
)

print(my_series["num3"])
print(my_series.iloc[0])
```

Look up a value using its assigned index.
Look up a value by its exact numerical position (0-based indexing), regardless of what the custom label is.

```python
# Series statistics

my_series = pd.Series([5, 8, 2, 11, 9])

print("min:", my_series.min())
print("max:", my_series.max())
print("mean:", my_series.mean())
print("median:", my_series.median())
```

Built-in methods to calculate the lowest value, highest value, average, and middle point of the data.

```python
# Import seaborn
import seaborn as sns

# Load Titanic dataset
titanic_data = sns.load_dataset("titanic")

titanic_data.head()
```

Displays the first N rows (default is 5).

```python
# Number of rows and columns

print(titanic_data.shape)
```

```python
# Select one column

age_column = titanic_data["age"]

print(type(age_column))

age_column.head()
```

Passing a single column name as a string returns that single column as a Series.

```python
# Select multiple columns

subset = titanic_data[["class", "sex", "age"]]

subset.head()
```

Displays the first N rows (default is 5).

```python
# Keep only male passengers

males = titanic_data[titanic_data["sex"] == "male"]

males.head()
```

creates a Boolean series that filters out the women and keeps only the men
Displays the first N rows (default is 5).

```python
# Keep male passengers traveling in First class

first_class_males = titanic_data[
    (titanic_data["sex"] == "male") &
    (titanic_data["class"] == "First")
]

first_class_males.head()
```

creates another Boolean series where only the male passengers that travelled in first class are kept
Displays the first N rows (default is 5).

```python
# Find passengers aged 20, 21, or 22

ages = [20, 21, 22]

young = titanic_data[
    titanic_data["age"].isin(ages)
]

young.head()
```

creates a list and checks if any of the passengers' ages match the values in the list
Displays the first N rows (default is 5).

```python
# Create a student DataFrame

scores = [
    {
        "Subject": "Mathematics",
        "Score": 85,
        "Grade": "B",
        "Remarks": "Good"
    },
    {
        "Subject": "History",
        "Score": 98,
        "Grade": "A",
        "Remarks": "Excellent"
    },
    {
        "Subject": "English",
        "Score": 76,
        "Grade": "C",
        "Remarks": "Fair"
    },
    {
        "Subject": "Science",
        "Score": 72,
        "Grade": "C",
        "Remarks": "Fair"
    },
    {
        "Subject": "Arts",
        "Score": 95,
        "Grade": "A",
        "Remarks": "Excellent"
    }
]

my_df = pd.DataFrame(
    scores,
    index=["Student1", "Student2", "Student3", "Student4", "Student5"]
)

my_df
```

creates a dataframe with a custom index

```python
# Select one row by label

my_df.loc["Student1"]
```

Extracts the row matching label "Student1" as a 1D Series

```python
# Select rows from Student1 through Student3

my_df.loc["Student1":"Student3"]
```

Slices rows between specified index labels inclusive (includes "Student3").

```python
# Select specific rows and columns

my_df.loc[
    "Student1":"Student4",
    ["Grade", "Score"]
]
```

takes rows from "Student1" to "Student4" while restricting columns to "Grade" and "Score".

```python
# Filter using .loc

my_df.loc[
    my_df["Score"] > 80,
    ["Score", "Grade"]
]
```

Evaluates a boolean row filter condition. Keeps rows where score is greater than 80 and returns only the "Score" and "Grade" columns.

```python
# Select the 4th row
# Position 3 means the 4th row because counting starts at 0

my_df.iloc[3]
```

selects the 4th row by its 0-based integer index position

```python
# Select rows at positions 2 and 3
# Select columns at positions 0 and 1

my_df.iloc[2:4, 0:2]
```

Slices row positions 2 to 3 (index 4 excluded) and column positions 0 to 1 (index 2 excluded).

```python
# Rebuild DataFrame with default index

my_df = pd.DataFrame(scores)

# Drop rows 1 and 4

smaller = my_df.drop(
    [1, 4],
    axis=0
)

smaller
```

Resets my_df back to default integer indices (0 to 4).

Returns a new DataFrame with rows at index positions 1 and 4 removed (axis=0 targets rows).

```python
# Drop columns

no_extras = my_df.drop(
    ["Subject", "Grade"],
    axis=1
)

no_extras
```

Returns a new DataFrame with "Subject" and "Grade" columns removed (axis=1 targets columns).

```python
# Sort passengers from youngest to oldest

by_age = titanic_data.sort_values(
    by="age"
)

by_age.head()
```

Sorts titanic_data by age from youngest to oldest (default ascending=True).

 Displays top 5 youngest passengers.



```python
# Sort passengers from oldest to youngest

by_age_desc = titanic_data.sort_values(
    by="age",
    ascending=False
)

by_age_desc.head()
```

Sorts titanic_data by age from oldest to youngest.

Displays top 5 oldest passengers.

```python
# Sort by class first and fare second

multi_sort = titanic_data.sort_values(
    by=["class", "fare"],
    ascending=[True, False]
)

multi_sort.head()
```

Sorts primary order by passenger class and secondary tie-breaker order by ticket fare.
Sorts class ascending and fare descending
Displays top 5 sorted results.
